```python
import threading

counter = []
lock = threading.Lock()

def count_up(n):
    global counter
    thread_id = threading.get_ident()  # 获取线程ID
    for _ in range(n):
        print(f"Thread {thread_id} finished counting.")
        # with lock:
        counter.append(1)  # 非原子操作

threads = []
for _ in range(3):
    t = threading.Thread(target=count_up, args=(100000,))
    t.start()
    threads.append(t)

for t in threads:
    t.join()

print("Expected:", 3 * 100000, "Actual:", len(counter))
```

```python
import multiprocess as mp
import os
import time

def count_up(n, counter, lock):
    pid = os.getpid()
    lock.acquire()  # 获取锁

    for _ in range(n):
        counter.value += 1
    print(f"Process {pid} finished counting.")
    
    lock.release()  # 释放锁

# counter = mp.Value('i', 0)
# count_up(2000, counter, mp.Lock())  # 测试单进程
# counter.value
start_time = time.time()

counter = mp.Value('i', 0)
lock = mp.Lock()
processes = []

for i in range(5):
    p = mp.Process(target=count_up, args=(10000, counter, lock))
    p.start()
    processes.append(p)

for p in processes:
    p.join()

print("Final counter value:", counter.value)
print("Elapsed time:", time.time() - start_time)
```
